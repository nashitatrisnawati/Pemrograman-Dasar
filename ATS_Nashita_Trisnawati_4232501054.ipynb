```python
#Nama : Nashita Trisnawati
#NIM : 4232501054
#Tanggal : 10 November 2026
#Mata Kuliah : Pemrograman Dasar (RPE311)
#Soal 9.Pembacaan Vibrasi dengan Sensor Rusak

data = ["1.2", "0.9", "ERR", "3.1", "1.8"]
jumlah_tidak_valid = 0

for n, nilai in enumerate(data, start=1):
    try:
        vibrasi = float(nilai)
    except ValueError:
        print(f"Unit {n}: data tidak valid")
        jumlah_tidak_valid += 1
        continue
    if vibrasi > 2.0:
        print(f"Warning: Unit {n} vibrasi tinggi!")
    if vibrasi > 3.0:
        print("KRITIS: pemeriksaan dihentikan")
        break

print("Jumlah data tidak valid:", jumlah_tidak_valid)
```

Program ini berisi pembacaan vibrasi dengan sensor rusak. terdsapat beberapa data yang dimasukkan akan tetpi tidak semua nya valid, program ini akan membaca mana data yang vallid dan mana yang tidak. dan juga mengkalisivikasi vibrasi

```python
#Nama : Nashita Trisnawati
#NIM : 4232501054
#Tanggal : 10 November 2026
#Mata Kuliah : Pemrograman Dasar (RPE311)
#CHALLENGE

data_konsumsi = ["11.5", "17.0", "25.5", "ERR", "7.0", "21.0"]

total_konsumsi = 0
jumlah_data_valid = 0
jumlah_data_tidak_valid = 0

print("Pemantauan Konsumsi Listrik Harian")

for hari, data in enumerate(data_konsumsi, start=1):
    try:
        konsumsi = float(data)
        print(f"\nHari ke-{hari}: {konsumsi} kWh")
        total_konsumsi += konsumsi
        jumlah_data_valid += 1

        if konsumsi < 10:
            print("Status: Konsumsi Rendah")
        elif konsumsi <= 20:
            print("Status: Konsumsi Normal")
        else:
            print("Status: Konsumsi Tinggi!")
            print("Peringatan: Periksa penggunaan listrik.")

    except ValueError:
        print(f"\nHari ke-{hari}: Data tidak valid!")
        jumlah_data_tidak_valid += 1
        continue

if jumlah_data_valid > 0:
    rata_rata = total_konsumsi / jumlah_data_valid
else:
    rata_rata = 0

print("Hasil Pemantauan")
print(f"Total konsumsi listrik : {total_konsumsi:.2f} kWh")
print(f"Rata-rata konsumsi : {rata_rata:.2f} kWh")
print(f"Data valid : {jumlah_data_valid}")
print(f"Data tidak valid : {jumlah_data_tidak_valid}")
```
