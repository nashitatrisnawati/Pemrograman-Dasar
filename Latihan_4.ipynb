```python
# Mennetukan tahun kabisat

tahun = int(input("Masukkan tahun :"))
if (tahun % 4 == 0 and tahun % 100 != 0) or (tahun % 400 == 0):
  print("tahun kabisat")
else :
  print("Bukan tahun kabisat")
```

```python
# Daya solar panel

tegangan = float(input("Masukkan tegangan solar panel (volt) : "))
arus = float(input("Masukkan arus solar panel (watt) : "))

daya = tegangan * arus

print("Daya solar panel :", (daya), "watt")
```
