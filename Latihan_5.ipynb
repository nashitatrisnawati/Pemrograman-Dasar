```python
#Belanja Buku

jumlah_buku = int(input("Masukkan jumlah buku : "))
harga_buku = int(input("Masukkan harga buku : "))
diskon = int(input("Masukkan diskon :"))
uang_bayar = int(input("Masukkan uang pembayaran : "))
jumlah_orang = int(input("Masukkan jumlah orang : "))

total_belanja = jumlah_buku * harga_buku
besar_diskon = total_belanja * diskon / 100
jumlah_bayar = total_belanja - besar_diskon

kembalian = uang_bayar - total_belanja

iuran_perorang = jumlah_bayar // jumlah_orang
sisa = jumlah_bayar % jumlah_orang

uang_cukup = uang_bayar >= jumlah_bayar

print(f"Total belanja = Rp{total_belanja}")
print(f"Besar diskon = Rp{besar_diskon}")
print(f"Jumlah yang harus dibayar = Rp{jumlah_bayar}")
print(f"Uang kembalian = Rp{kembalian}")
print(f"Iuran per orang = Rp{iuran_perorang}")
print(f"Sisa = Rp{sisa}")
print(f"Apakah uangnya cukup? {uang_cukup}")
```
