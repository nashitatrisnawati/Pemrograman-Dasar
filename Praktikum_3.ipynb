```python
#B1.1

a = int (input("Masukkan bilangan pertama: "))
b = int (input("Masukkan bilangan Kedua: "))

print("Hasil Penjumlahan =", a+b)
print("Hasil Pengurangan =", a-b)
print("Hasil Perkalian =", a*b)
print("Hasil Pembagian =", a/b)
print("Hasil Pembagian bulat =", a//b)
print("Hasil Sisa pembagian =", a%b)
```

```python
# B1.2

nilai = int(input("Masukkan nilai ujian: "))

print("Nilai ini termasuk kategori lulus:", nilai >= 65)
```

```python
#B1.3

umur = int(input("Masukkan umur: "))
punya_sim = input("Apakah memiliki SIM? (True/False): ") == "True"

boleh_mengemudi = umur >= 17 and punya_sim

print("Anda bolhe mengemudi sendiri:", boleh_mengemudi)
```

```python
#B2.4

energi_input = float(input("Masukkan energi input (kWh): "))
energi_output = float(input("Masukkan energi output (kWh): "))

efisiensi = (energi_output / energi_input) * 100

print(f"Efiseinsi listrik = {efisiensi:.2f}%")
```

```python
#B2.5

suhu_boiler = float(input("Masukka suhu boiler (°C): "))
beban_operasi = float(input("MAsukkan beban operasi (%): "))

perlu_maintenance = (suhu_boiler > 500) or (beban_operasi > 90)

print(f"Apakah unit perlu mantenance? {perlu_maintenance}")
```
